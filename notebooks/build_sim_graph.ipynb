```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
uldf = pd.read_csv("../sim_matrix/unigram_lex_matrix.csv", index_col=0)
bldf = pd.read_csv("../sim_matrix/bigram_lex_matrix.csv", index_col=0)
tldf = pd.read_csv("../sim_matrix/trigram_lex_matrix.csv", index_col=0)

uwdf = pd.read_csv("../sim_matrix/unigram_word_matrix.csv", index_col=0)
bwdf = pd.read_csv("../sim_matrix/bigram_word_matrix.csv", index_col=0)
twdf = pd.read_csv("../sim_matrix/trigram_word_matrix.csv", index_col=0)
```

```python
# Create a heatmap
plt.figure(figsize=(12, 10))
sns.heatmap(uldf, annot=True, fmt=".2f", cmap="RdYlGn", cbar=True)
plt.title("Language Unigram Lexical-level Similarity Matrix (Dice Coefficient)")
plt.xticks(rotation=45)
plt.yticks(rotation=0)
plt.tight_layout()
plt.savefig("../output/lexical_sim/unigram_letter_similarity_heatmap.png")
plt.show()
```

```python
plt.figure(figsize=(12, 10))
sns.heatmap(bldf, annot=True, fmt=".2f", cmap="RdYlGn", cbar=True)
plt.title("Language Bigram Lexical-level Similarity Matrix (Dice Coefficient)")
plt.xticks(rotation=45)
plt.yticks(rotation=0)
plt.tight_layout()
plt.savefig("../output/lexical_sim/bigram_letter_similarity_heatmap.png")
plt.show()
```

```python
plt.figure(figsize=(12, 10))
sns.heatmap(tldf, annot=True, fmt=".2f", cmap="RdYlGn", cbar=True)
plt.title("Language Trigram Lexical-level Similarity Matrix (Dice Coefficient)")
plt.xticks(rotation=45)
plt.yticks(rotation=0)
plt.tight_layout()
plt.savefig("../output/lexical_sim/trigram_letter_similarity_heatmap.png")
plt.show()
```

```python
plt.figure(figsize=(12, 10))
sns.heatmap(uwdf, annot=True, fmt=".2f", cmap="RdYlGn", cbar=True)
plt.title("Language Trigram Token-level Similarity Matrix on ngram-count >= 2 (Dice Coefficient)")
plt.xticks(rotation=45)
plt.yticks(rotation=0)
plt.tight_layout()
plt.savefig("../output/token_sim/unigram_word_similarity_heatmap.png")
plt.show()
```
