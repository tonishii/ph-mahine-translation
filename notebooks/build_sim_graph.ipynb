```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../similarity/matrix.csv", index_col=0)
```

```python
# Create a heatmap
plt.figure(figsize=(12, 10))
sns.heatmap(df, annot=True, fmt=".2f", cmap="RdYlGn", cbar=True)
plt.title("Language Similarity Matrix (Dice Coefficient)")
plt.xticks(rotation=45)
plt.yticks(rotation=0)
plt.tight_layout()
plt.savefig("../output/similarity_heatmap.png")
plt.show()
```
