```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
udf = pd.read_csv("../sim_matrix/unigram_matrix.csv", index_col=0)
bdf = pd.read_csv("../sim_matrix/bigram_matrix.csv", index_col=0)
tdf = pd.read_csv("../sim_matrix/trigram_matrix.csv", index_col=0)
```

```python
# Create a heatmap
plt.figure(figsize=(12, 10))
sns.heatmap(udf, annot=True, fmt=".2f", cmap="RdYlGn", cbar=True)
plt.title("Language Unigram Lexical-level Similarity Matrix (Dice Coefficient)")
plt.xticks(rotation=45)
plt.yticks(rotation=0)
plt.tight_layout()
plt.savefig("../output/lexical_sim/unigram_similarity_heatmap.png")
plt.show()
```

```python
plt.figure(figsize=(12, 10))
sns.heatmap(bdf, annot=True, fmt=".2f", cmap="RdYlGn", cbar=True)
plt.title("Language Bigram Lexical-level Similarity Matrix (Dice Coefficient)")
plt.xticks(rotation=45)
plt.yticks(rotation=0)
plt.tight_layout()
plt.savefig("../output/lexical_sim/bigram_similarity_heatmap.png")
plt.show()
```

```python
plt.figure(figsize=(12, 10))
sns.heatmap(tdf, annot=True, fmt=".2f", cmap="RdYlGn", cbar=True)
plt.title("Language Trigram Lexical-level Similarity Matrix (Dice Coefficient)")
plt.xticks(rotation=45)
plt.yticks(rotation=0)
plt.tight_layout()
plt.savefig("../output/lexical_sim/trigram_similarity_heatmap.png")
plt.show()
```
