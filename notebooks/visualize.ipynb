```python
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from get_bleu_scores import read_bleu_score
```

```python
dfs = read_bleu_score()
```

```python
for i, df in enumerate(dfs):
    
    scalar = MinMaxScaler()
    # Use underscore consistently
    df['score_normalized'] = scalar.fit_transform(df[['score']])
    
    plt.figure(figsize=(8,5))
    plt.plot(df['step'], df['score_normalized'], marker='o', label='Normalized BLEU Score')
    
    plt.xlabel("Training Step")
    plt.ylabel("BLEU Score")
    plt.title(f"BLEU Score Progress - {getattr(df, 'name')}")
    plt.grid(True)
    plt.legend()
    plt.show()
    
    plt.figure(figsize=(8,5))
    plt.bar(df['step'], df['score'], color='skyblue')
    plt.xlabel("Training Step")
    plt.ylabel("BLEU Score")
    plt.title(f"Actual BLEU Scores - {getattr(df, 'name')}")
    plt.grid(axis='y')
    plt.show()
```
